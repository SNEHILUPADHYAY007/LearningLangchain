```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI       
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
load_dotenv(override = True)

google_api_key = os.getenv("GOOGLE_API_KEY")

if not google_api_key:
    print("API key not found")
else:
    print("API Key found")

chat = ChatGoogleGenerativeAI(model = "gemini-2.5-flash")
```

## INVOKE vs GENERATE(with the usecase)

If I have multiple independent tasks, I can run them in parallel using generate().
Later, based on my use-case, I can take the outputs and feed them one-by-one into invoke() for reasoning or decision-making. Each tasks within generate() is a simple independent task in itself.




```python
# Using Invoke
result = chat.invoke([
    SystemMessage(content="You're an expert Python Programmer and doesn't know anything else about other programming language"),
    HumanMessage(content="Help me understand about Scala val and var keywords")
])

print(result.content)
```

```python
# Using Generate 
# (Each prompt will act as an independent task)
# (Since tasks are independent, System prompts should be different)

result = chat.generate([[
        SystemMessage(content="You're an expert Python Programmer and doesn't know anything else about other programming language"),
        HumanMessage(content="Help me understand about Scala val and var keywords")
    ],
    [
        SystemMessage(content="You're an expert Python Programmer and doesn't know anything else about other programming language"),
        HumanMessage(content="Help me understand about Scala val and var keywords")
    ]])

print(result.generations)
```

```python
# Model parameters:- max_output_tokens, temperature, top-k, top-p
# Memory - InMemoryCache, set_llm_cache

chat = ChatGoogleGenerativeAI(model = "gemini-2.5-flash-lite", 
                              max_tokens= 250,
                              temperature = 1,
                              top_k=5,
                              top_p=0.7)
result = chat.invoke([
    SystemMessage(content="You're an expert Python Programmer and doesn't know anything else about other programming language"),
    HumanMessage(content="Help me understand about Scala val and var keywords")
])

print(result.content)
```
