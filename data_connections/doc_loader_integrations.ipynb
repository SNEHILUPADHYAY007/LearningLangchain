```python
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader("https://news.ycombinator.com/item?id=34817881").load()

print(loader[0].page_content)
```

```python
from langchain_core.prompts import HumanMessagePromptTemplate, ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI

human_msg = HumanMessagePromptTemplate.from_template("Summarise the Given Website Content into Super simpler understandable format:\n{content}")

chat_msg_template = ChatPromptTemplate.from_messages([human_msg])
final_prompt = chat_msg_template.format_prompt(content=loader[0].page_content).to_messages()

chat = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")

res = chat.invoke(final_prompt)
print(res.content)
```
