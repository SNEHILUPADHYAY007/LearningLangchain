```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, SystemMessage

env_path = '../.env'
```

### Initializing the Google Model

```python
load_dotenv(override=True, dotenv_path=env_path)

google_api_key = os.getenv("GOOGLE_API_KEY")a

if google_api_key:
    llm = ChatGoogleGenerativeAI(
        model="gemini-2.5-flash",
    )
    print("ChatGoogleGenerativeAI LLM initialized.")    
else:
    print("Failed to load API Key.")
```

### Using Chat Completion Model functionality of Google
1) .invoke:- Use .invoke():

-Single prompt at a time
-Conversational flows (with memory/chat models)
-Most common use case

2) .generate:- Use .generate(): will send the request in batches(NOT ASYNC/PARALLEL)

-Multiple independent prompts (no conversation needed)
-Example: Summarizing 10 different articles
-When you need metadata (token counts, etc.)

```python
message = llm.invoke([
    SystemMessage(content="You are a funny translator."),
    HumanMessage(content="Translate the following English text to French: 'Hello, how are you?'")
])

# print(message.content)

message_2 = llm.generate([
    [
        SystemMessage(content="You are a funny translator."),
        HumanMessage(content="Translate the following English text to French: 'What is your name?'")
    ],
    [
        SystemMessage(content="You are a funny translator."),
        HumanMessage(content="Translate the following English text to French: 'Where is the nearest restaurant?'")
    ]
])

print(message_2.generations[0][0].text)
```
