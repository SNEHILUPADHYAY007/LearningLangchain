### Initializing Google Model

top_p (also called nucleus sampling) controls the randomness of the model's output by limiting which tokens it can choose from.
How it works:
Instead of considering ALL possible next tokens, the model only considers the smallest set of tokens whose cumulative probability adds up to p.
Visual Example:
Let's say the model is predicting the next word after "The cat is very":
Possible next words (with probabilities):
cute:    40% (0.40)
fluffy:  30% (0.30)
small:   15% (0.15)
tiny:    10% (0.10)
large:    3% (0.03)
angry:    2% (0.02)
With top_p = 0.9:
1. Sort by probability (already sorted above)
2. Add probabilities until reaching 90%:
   - cute: 40% → cumulative: 40%
   - fluffy: 30% → cumulative: 70%
   - small: 15% → cumulative: 85%
   - tiny: 10% → cumulative: 95% ✓ (reached 90%)

3. Only consider: cute, fluffy, small, tiny
4. Ignore: large, angry (low probability tokens cut off)
With top_p = 0.5:
- cute: 40% → cumulative: 40%
- fluffy: 30% → cumulative: 70% ✓ (reached 50%)

Only consider: cute, fluffy

```python
load_dotenv(override=True, dotenv_path=env_path)
api_key = os.getenv("GOOGLE_API_KEY")

if api_key:
    chat_llm = ChatGoogleGenerativeAI(
        model="gemini-2.5-flash-lite",
        temperature=1,
        # max_output_tokens=500,
        # top_k = 50,
        # top_p = 1
    )
    print("ChatGoogleGenerativeAI LLM initialized.")
else:
    print("Failed to load API Key.")
```

```python
print(chat_llm.invoke("Best snacks item to have").content)
```

### Caching
Package:- langchain_core.caches 

```python
from langchain_core.caches import InMemoryCache
from langchain_core.globals import set_llm_cache

set_llm_cache(InMemoryCache())

result = chat_llm.invoke("Tell me about Mts.")
print("*" * 50)
print(result.content)

result = chat_llm.invoke("Tell me about Mts.")
print("*" * 50)
print(result.content)

result = chat_llm.invoke("Tell me about Mts.")
print("*" * 50)
print(result.content)
```
