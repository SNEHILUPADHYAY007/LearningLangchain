```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import PromptTemplate
from langchain_core.prompts import SystemMessagePromptTemplate, HumanMessagePromptTemplate, AIMessagePromptTemplate, ChatPromptTemplate

env_path = '../.env'
load_dotenv(dotenv_path=env_path)
```

```python
api_key = os.getenv("GOOGLE_API_KEY")

if api_key: 
    chat_llm = ChatGoogleGenerativeAI(
        model="gemini-2.5-flash-lite",
        temperature=1,
        # max_output_tokens=500,
        # top_k = 50,
    )
    print("ChatGoogleGenerativeAI LLM initialized.")

else:
    print("GOOGLE_API_KEY not found in environment variables.") 
```

### Using PromptTemplate

```python
prompt = PromptTemplate(
    input_variables=["topic", "level"],
    template="Provide me with three detailed points on the topic of {topic}. It should be for {level}"
)
chat_llm.invoke(prompt.format(topic="Artificial Intelligence", level="beginner")).content
```

### Using ChatPromptTemplate

```python
system_message = "You are an expert {topic}, who excels in making food in under {level} having less toxic items."

sys_msg_prompt = SystemMessagePromptTemplate.from_template(system_message)
human_msg_prompt = HumanMessagePromptTemplate.from_template("Suggest me a recipe for {dish_name}.")

chat_prompt = ChatPromptTemplate.from_messages([sys_msg_prompt, human_msg_prompt])

formatted_prompt = chat_prompt.format_prompt(topic="chef", level="30 minutes", dish_name="pasta")

response = chat_llm.invoke(formatted_prompt.to_messages())
print(response.content)
```

### Few-Shot Prompting

Here, Lots of Few Shot examples can be created using the below approach, Make sure to keep tokens in mind.

```python
sys_msg = "You're an expert in {role} and capable of simplifying complex topics into easy-to-understand points. You can refer the examples provided in HumanMessage and AIMessage to get an idea of how to respond."

sys_msg_prompt = SystemMessagePromptTemplate.from_template(sys_msg)

human_text = "Within thermodynamics, the irreversible interplay between entropy production, non equilibrium constraints, and microscopic fluctuations governs how macroscopic systems exchange heat, perform work, and evolve toward equilibrium, revealing that energy conservation alone cannot predict directionality, efficiency limits, or spontaneous processes without statistical interpretation across classical, quantum, and cosmological physical regimes universally."

human_msg_prompt = HumanMessagePromptTemplate.from_template(human_text)

ai_text = "Thermodynamics explains how energy and heat move and why natural processes happen in one direction rather than reversing on their own."

ai_msg_prompt = AIMessagePromptTemplate.from_template(ai_text)

chat_prompt = ChatPromptTemplate.from_messages([sys_msg_prompt, human_msg_prompt,ai_msg_prompt])
final_prompt = chat_prompt.format_prompt(role="thermodynamics") 
print(final_prompt)
# response = chat_llm.invoke(final_prompt.to_messages())
# print(response.content)
```
