```python
import os
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, HumanMessagePromptTemplate, SystemMessagePromptTemplate
from langchain_core.globals import set_llm_cache
from langchain_core.caches import InMemoryCache

env_path = '../.env'
load_dotenv(dotenv_path=env_path)

api_key = os.getenv("GOOGLE_API_KEY")
if api_key:
    chat = ChatGoogleGenerativeAI(
        model="gemini-2.5-flash-lite",
        temperature=0.7,
        max_output_tokens=500,
        top_k = 40,
    )
    print("GoogleGenerativeAI LLM initialized.")
else:
    print("GOOGLE_GENAI_API_KEY not found in environment variables.")
```

We can still not be able to get required format as we'll get string from LLM and we have to think of some methods to further correct it and make it as per our requirement.

for eg:- llm_output = 'red', 'green', 'yellow'
and when we'll do list(llm_output) it'll create individual characters only. So we have to utilize this output parser to work in this case and get required result. Check Cell-3

```python
# Step-1 import, initialization
from langchain_core.output_parsers import  CommaSeparatedListOutputParser
output_parser = CommaSeparatedListOutputParser()
set_llm_cache(InMemoryCache())

# We can also see what is the default instn
output_parser.get_format_instructions()

# \n to segreated the info that need to be passed to LLM
human_msg = "{request}\n{format_instn}"

# String msg to template object
human_msg_template = HumanMessagePromptTemplate.from_template(human_msg)

# Grouping all the msg available
chat_template = ChatPromptTemplate.from_messages([human_msg_template])

final_chat = chat_template.format_prompt(
    request="Provide me with three random characteristics of LLM.",
    format_instn=output_parser.get_format_instructions()
).to_messages()

resp = chat.invoke(final_chat).content

# Formating into list
print(output_parser.parse(resp))
```

```python
# Example to above Markdown:-
llm_output = "red, green, yellow"
# Converting it to list using list(llm_output) will break the requirement.

output_parser.parse(llm_output)
```
